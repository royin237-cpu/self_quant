电脑上要有同花顺


```python
import easytrader
from easytrader import grid_strategies
import os
import tempfile

# 创建可控的临时文件目录（避免系统 temp 目录权限/清理问题导致 .xls 文件丢失）
tmp_dir = os.path.join(os.path.expanduser('~'), 'easytrader_tmp')
os.makedirs(tmp_dir, exist_ok=True)

# 1. 实例化客户端
# 'universal_client' 专门用于适配同花顺通用版客户端
user = easytrader.use('universal_client')

# 设置 grid_strategy 为 Xls，并指定 tmp_folder 到可控目录
user.grid_strategy = grid_strategies.Xls(tmp_folder=tmp_dir)

# 2. 连接同花顺客户端
# 注意：这里要填入你电脑上同花顺下单程序（xiadan.exe）的准确物理路径
user.prepare(exe_path=r'C:\1_software\tonghuashun\同花顺\xiadan.exe')

# 3. 获取账户信息（测试是否连接成功）
try:
    balance = user.balance
    print("资金情况：", balance)

    positions = user.position
    print("当前持仓：", positions)
except Exception as e:
    print("连接失败，请检查客户端是否在最前端并已登录。错误信息：", e)
```

```python
# 假设你的策略得出要买入平安银行（000001），价格10.5元，数量100股
try:
    user.enable_type_keys_for_editor()
except Exception as e:
    print(f"enable_type_keys 警告（可忽略）: {e}")

stock_code = '600477'
target_price = 3.24
amount_to_buy = 1000

print(f"准备买入 {stock_code}...")

# 执行买入（注意：实盘测试时务必先用100股便宜的股票测试！）
try:
    buy_receipt = user.buy(stock_code, price=target_price, amount=amount_to_buy)
    print("买入回执:", buy_receipt)
except Exception as e:
    print(f"买入失败: {e}")
    print("请检查：1)同花顺客户端是否在最前端 2)是否已登录 3)股票代码/价格是否有效")

# 执行卖出
# sell_receipt = user.sell('000001', price=11.0, amount=100)

# 撤单（如果需要的话，传入委托单号或一键撤销所有）
# user.cancel_entrust('委托单号')
```
